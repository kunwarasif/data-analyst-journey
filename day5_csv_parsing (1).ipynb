```python
def get_total_sales(data):
    total = 0
    for sale in data:
        total = total + sale["amount"]
    return total
```

```python
def get_transaction_count(data):
    count = len(data)
    return count
```

```python
def get_average_sale(data):
    average = get_total_sales(data)/get_transaction_count(data)
    return average
```

```python
def get_sales_by_person(data):
    totals = {}
    for sales in data:
        name = sales["salesperson"]
        amount = sales["amount"]
        if name in totals:
            totals[name] = totals[name] + sales["amount"]
        else:
            totals[name] = sales["amount"]
    return totals
```

```python
def get_sales_by_month(data):
    totals = {}
    for sales in data:
        month = sales["month"]
        amount = sales["amount"]
        if month in totals:
            totals[month] = totals[month] + sales["amount"]
        else:
            totals[month] = sales["amount"]
    return totals
```

```python
def get_top_performer(data):
    top_performer = ""
    sales_total = get_sales_by_person(data)
    amount = 0
    for sales_person, totals in sales_total.items():
        if totals > amount:
            top_performer = sales_person
            amount = totals
    return (top_performer, amount)
    
```

```python
def generate_report(data):
    """
    Generate a formatted sales report from a list of sale records.
    
    Args:
        data: list of dicts, each with keys 'salesperson', 'amount', 'month'
    
    Prints:
        A formatted multi-section report to the console.
    """
    # Section 1: Header
    print("=" * 50)
    print("       MONTHLY SALES REPORT")
    print("=" * 50)
    print()
    
    # Section 2: Summary stats
    total = get_total_sales(data)
    count = get_transaction_count(data)
    average = get_average_sale(data)
    
    print(f"Total Sales: ₹{total:,}")
    print(f"Number of Transactions: {count}")
    print(f"Average Sale Value: ₹{average:,.0f}")
    print()
    # Section 3: Sales by Person
    print("--- Sales by Person ---")
    
    person_totals = get_sales_by_person(data)
    sorted_persons = sorted(person_totals.items(), key=lambda x: x[1], reverse=True)
    
    for name, amount in sorted_persons:
        percentage = amount / total
        print(f"{name:<10} ₹{amount:>10,}   ({percentage:.1%})")
    
    print()

    # Section 4: Sales by Month
    print("--- Sales by Month ---")
    month_totals = get_sales_by_month(data)
    sorted_month = sorted(month_totals.items(), key=lambda x: x[1], reverse=True)

    for month, amount in sorted_month:
        print(f"{month:<10} ₹{amount:>10,}")

    print()

    # Section 5: Top Performer
    print("--- Top Performer ---")
    top_name, top_amount = get_top_performer(data)
    print(f"{top_name} with ₹{top_amount:,}")
    print() 

    # Section 5: Big Sale
    print("--- Big Sales (>₹60,000) ---")
    big_seller = set()
    for sale in data:
        if sale["amount"] > 60000:
            big_seller.add(sale["salesperson"])
    print(", ".join(big_seller))
    print() 

    # Section 7: Footer
    print("=" * 50)
    
```

```python
def is_valid_record(record):
    """Check if a sales record meets all validation rules."""
    
    # Rule 1: salesperson must be non-empty
    if not record.get("salesperson") or not str(record["salesperson"]).strip():
        return False
    # Rule 2: amount must be a number (not None, not string)
    if not isinstance(record.get("amount"), (int, float)):
        return False
    # Rule 3: amount must be positive
    if record["amount"] <= 0:
        return False
    # Rule 4: month must be non-empty
    if not record.get("month") or not str(record["month"]).strip():
        return False
    # Rule 5: month must be a valid month
    valid_months = {"Jan", "Feb", "Mar", "Apr", "May", "Jun", 
                    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"}
    if record["month"].strip() not in valid_months:
        return False

    return True 
```
