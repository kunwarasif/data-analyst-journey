```python
df["year"] = df["sale_date"].dt.year
print(df[["sale_date", "year"]])
```

```python
# Month number (1-12)
df["month_num"] = df["sale_date"].dt.month

# Month name
df["month_name"] = df["sale_date"].dt.month_name()

# Day of month (1-31)
df["day"] = df["sale_date"].dt.day

# Day of week — name
df["weekday_name"] = df["sale_date"].dt.day_name()

# Day of week — number (Monday=0, Sunday=6)
df["weekday_num"] = df["sale_date"].dt.weekday

# Quarter (1-4)
df["quarter"] = df["sale_date"].dt.quarter

# Day of year (1-365)
df["day_of_year"] = df["sale_date"].dt.dayofyear

print(df[["sale_date", "year", "month_name", "weekday_name", "quarter"]])
```

```python
# Add 30 days
df["warranty_end"] = df["sale_date"] + pd.Timedelta(days=30)
```

```python
# Q1: How many days between each sale and the most recent sale?
df["days_since_sale"] = (df["sale_date"].max() - df["sale_date"]).dt.days
print(df[["sale_date", "days_since_sale"]])
```

```python
# Q2: Filter to sales in the second half of 2024 (July onwards)
df[df["sale_date"] >= "2024-07-01"]
```

```python
# Q3: For each salesperson, what's the date range of their sales?
df.groupby("salesperson")["sale_date"].agg(["min", "max"])
```

```python
# Q4: For each salesperson, how many days from first to last sale?
span = df.groupby("salesperson")["sale_date"].agg(["min", "max"])
span["span_days"] = (span["max"] - span["min"]).dt.days
print(span)
```

```python
# Q5: Add a "next quarter check-in" column — 90 days after each sale
df["check_in_date"] = df["sale_date"] + pd.Timedelta(days=90)
print(df[["sale_date", "check_in_date"]])
```

```python
today = pd.Timestamp("2024-12-31")
six_months_ago = today - pd.DateOffset(months=6)

weekend_recent = df[
    (df["sale_date"] >= six_months_ago) &     # last 6 months
    (df["sale_date"].dt.weekday >= 5)         # weekends only (Sat=5, Sun=6)
]

print(weekend_recent["amount"].mean())
print(f"\nNumber of weekend sales in last 6 months: {len(weekend_recent)}")
```

```python
df["is_weekend"] = df["sale_date"].dt.weekday >= 5
last_6_months = df[df["sale_date"] >= six_months_ago]

last_6_months.groupby("is_weekend")["amount"].agg(["mean", "count", "sum"])
```
