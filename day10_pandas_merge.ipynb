```python
import pandas as pd

# Create the sales file
sales_csv = """sale_id,salesperson_id,amount,month
1,101,50000,Jan
2,102,75000,Jan
3,101,60000,Feb
4,103,80000,Feb
5,102,45000,Feb
6,103,90000,Mar
7,101,55000,Mar"""

with open("sales_v2.csv", "w") as f:
    f.write(sales_csv)

# Create the employees file
employees_csv = """salesperson_id,name,city,commission_rate
101,Rohit,Delhi,0.05
102,Amit,Mumbai,0.06
103,Neha,Bangalore,0.05
104,Vikram,Chennai,0.04"""

with open("employees.csv", "w") as f:
    f.write(employees_csv)

# Load both
sales_df = pd.read_csv("sales_v2.csv")
employees_df = pd.read_csv("employees.csv")

print("--- sales_df ---")
print(sales_df)
print("\n--- employees_df ---")
print(employees_df)
```

```python
merged = pd.merge(sales_df, employees_df, on= "salesperson_id", how = "inner")
print(merged)
```

```python
sales_df.merge(employees_df, on= "salesperson_id", how = "inner")
```

```python
#If sales_df had a column called emp_id and employees_df had salesperson_id
#pd.merge(sales_df, employees_df, left_on="emp_id", right_on="salesperson_id", how="inner")
```

```python
sales_df.merge(employees_df, on ="salesperson_id", how = "inner").groupby("city")["amount"].sum()
```

```python
# Q1: Inner join — how many rows in the result?
result = sales_df.merge(employees_df, on= "salesperson_id", how = "inner")
print(result)
print(f"\nShape: {result.shape}")
```

```python
# Q2: After joining, what's the total sales by city?
print(result.groupby("city")["amount"].sum())
```

```python
# Q3: What's the commission for each sale? 
# (multiply amount by commission_rate, store in a new column)
result["commission"] = result["amount"] * result["commission_rate"]
print(result[["name", "amount", "commission_rate", "commission"]])
```

```python
# Q4: What's each salesperson's total commission?
print(result.groupby("name")["commission"].sum())
```

```python
# Q1: Left join — same as inner here?
left_result = sales_df.merge(employees_df, on = "salesperson_id", how = "left")
print(left_result.shape)
print(left_result)
```

```python
# Q2: Right join — how many rows?
right_result = sales_df.merge(employees_df, on="salesperson_id", how="right")
print(right_result.shape)
print(right_result)
```

```python
# Q3: Outer join — how many rows?
outer_result = sales_df.merge(employees_df, on="salesperson_id", how="outer")
print(outer_result.shape)
print(outer_result)
```

```python
# Q4: After right join, can we find employees with NO sales?
# Hint: filter for rows where sale_id is NaN
print(right_result[right_result["sale_id"].isna()])
```

```python
# Add product info to each sale
sales_with_products_csv = """sale_id,salesperson_id,product_id,amount,month
1,101,P01,50000,Jan
2,102,P02,75000,Jan
3,101,P03,60000,Feb
4,103,P01,80000,Feb
5,102,P02,45000,Feb
6,103,P03,90000,Mar
7,101,P01,55000,Mar"""

with open("sales_v3.csv", "w") as f:
    f.write(sales_with_products_csv)

# Create products lookup
products_csv = """product_id,product_name,category
P01,Laptop,Electronics
P02,Phone,Electronics
P03,Office Chair,Furniture"""

with open("products.csv", "w") as f:
    f.write(products_csv)

sales_df = pd.read_csv("sales_v3.csv")
products_df = pd.read_csv("products.csv")

print("--- sales_df ---")
print(sales_df)
print("\n--- products_df ---")
print(products_df)
print("\n--- employees_df (unchanged) ---")
print(employees_df)
```

```python
full_view = sales_df.merge(employees_df, on="salesperson_id").merge(products_df, on="product_id")
print(full_view)
```

```python
# Q1: Sales by product category
full_view.groupby("category")["amount"].sum()
```

```python
# Q2: Sales by city + category (cross-tab)
full_view.groupby(["city", "category"])["amount"].sum()
```

```python
# Q3: Top-selling product
full_view.groupby("product_name")["amount"].sum().idxmax()
```

```python
# Q4: Each salesperson's biggest deal — and what they sold
full_view.sort_values("amount", ascending=False).groupby("name").first()
```

```python
# Q1 — Total sales by category
full_view.groupby("category")["amount"].sum()
```

```python
# Q2 — Total sales by category AND city
full_view.groupby(["category", "city"])["amount"].sum()
```

```python
# Q3 — Which salesperson sold the most "Electronics"?
electronics = full_view[full_view["category"] == "Electronics"]
print(electronics.groupby("name")["amount"].sum().sort_values(ascending=False))
```

```python
# Q4 — Average sale amount per product
full_view.groupby("product_name")["amount"].mean()
```

```python
# Q5 — Commission earned per product
full_view["commission"] = full_view["amount"] * full_view["commission_rate"]
full_view.groupby("product_name")["commission"].sum()
```

```python
electronics = full_view[full_view["category"] == "Electronics"]
print(electronics.groupby("name")["amount"].sum().sort_values(ascending=False))
```
