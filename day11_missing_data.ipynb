```python
import pandas as pd
import numpy as np

# A messy customer dataset
messy_data = {
    "customer_id": [1, 2, 3, 4, 5, 6, 7, 8],
    "name": ["Rohit", "Amit", "Neha", "Vikram", None, "Priya", "Suresh", None],
    "age": [28, 35, np.nan, 45, 30, np.nan, 52, 41],
    "city": ["Delhi", "Mumbai", "Bangalore", None, "Chennai", "Delhi", None, "Mumbai"],
    "purchase_amount": [50000, 75000, np.nan, 45000, 60000, 80000, np.nan, 55000],
    "last_purchase_date": ["2024-01-15", "2024-02-20", None, "2024-03-10", "2024-02-05", None, "2024-01-30", "2024-03-25"]
}

df = pd.DataFrame(messy_data)
print(df)
```

```python
# True/False mask of missing cells across the whole DataFrame
df.isna()
```

```python
# Per-column count of missing values
df.isna().sum()
```

```python
# Total missing cells in the whole DataFrame
df.isna().sum().sum()
```

```python
# Percentage missing per column
(df.isna().sum()/ len(df)) * 100
```

```python
# Rows that have ANY missing values
df[df.isna().any(axis = 1)]
```

```python
# Rows that have ALL values missing (complete junk rows)
df[df.isna().all(axis = 1)]
```

```python
# Columns that have ANY missing values
df.columns[df.isna().any()]
```

```python
# Opposite — completely clean rows
df[df.notna().all(axis = 1)]
```

```python
df.isna().sum()              # axis=0 default — sum down rows = count per column
```

```python
df.isna().sum(axis=1)        # sum across columns = count per row
```

```python
# Q1: How many missing values per column?
df.isna().sum()
```

```python
# Q2: How many missing values total in the DataFrame?
df.isna().sum().sum()
```

```python
# Q3: Percentage missing per column
(df.isna().sum()/ len(df) ) *100
```

```python
# Q4: Show only rows that have at least one missing value
df[df.isna().any(axis = 1)]
```

```python
# Q5: Show only rows where ALL values are present
df[df.notna().all(axis = 1)]
```

```python
# Q6: How many missing values in each row?
df.isna().sum(axis=1)
```

```python
# Total rows
print(f"Total rows: {len(df)}")

# Rows with any missing value
missing_rows = df.isna().any(axis=1).sum()
print(f"Rows with any missing: {missing_rows}")

# Percentage that would be lost with dropna()
pct_lost = (missing_rows / len(df)) * 100
print(f"Percentage lost if we dropna(): {pct_lost}%")
```

```python
# Just to see what dropna does mechanically
print("Original rows:", len(df))
print("After dropna():", len(df.dropna()))
print("After dropna(subset=['age']):", len(df.dropna(subset=["age"])))
print("After dropna(subset=['age', 'city']):", len(df.dropna(subset=["age", "city"])))
print("After dropna(thresh=5):", len(df.dropna(thresh=5)))
```

```python
# Strategy 1: constant
print("--- Fill city with 'Unknown' ---")
print(df["city"].fillna("Unknown"))
```

```python
# Strategy 2: zero (let's see if this is appropriate or not)
print("\n--- Fill purchase_amount with 0 ---")
print(df["purchase_amount"].fillna(0))
```

```python
# Strategy 3: mean
print("\n--- Fill age with mean ---")
print(df["age"].fillna(df["age"].mean()))
```

```python
# Strategy 3b: median
print("\n--- Fill age with median ---")
print(df["age"].fillna(df["age"].median()))
```

```python
# Strategy 4: mode (most common)
print("\n--- Fill city with mode ---")
print(df["city"].fillna(df["city"].mode()[0]))
```

```python
df["age"].fillna(df["age"].mean())   # returns new Series, df unchanged
```
