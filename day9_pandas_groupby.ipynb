```python
def quick_report(df):
    print(f"Total Sales: ₹{df['amount'].sum():,}")
    print(f"Number of Transactions: {len(df)}")
    print(f"Average Sale Value: ₹{df['amount'].mean():,.0f}")
    
    print(f"\n--- Sales by Person (sorted) ---")
    print(df.groupby("salesperson")["amount"].sum().sort_values(ascending=False))
    
    print(f"\n--- Sales by Month (sorted) ---")
    print(df.groupby("month")["amount"].sum().sort_values(ascending=False))
    
    print(f"\n--- Top Performer ---")
    top_name = df.groupby("salesperson")["amount"].sum().idxmax()
    top_amount = df.groupby("salesperson")["amount"].sum().max()
    print(f"{top_name} with ₹{top_amount:,}")
    
    print(f"\n--- Big Sales (>₹60,000) ---")
    big_sellers = df[df["amount"] > 60000]["salesperson"].unique()
    print(", ".join(big_sellers))

    
    print(f"\n--- Sales Percentiles ---")
    print(f"25th: ₹{df['amount'].quantile(0.25):,.0f}")
    print(f"50th (median): ₹{df['amount'].quantile(0.50):,.0f}")
    print(f"75th: ₹{df['amount'].quantile(0.75):,.0f}")

quick_report(df)
```
