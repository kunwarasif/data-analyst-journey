```python
sales_data = [
    {"salesperson": "Rohit", "amount": 50000, "month": "Jan"},
    {"salesperson": "Amit", "amount": 75000, "month": "Jan"},
    {"salesperson": "Rohit", "amount": 60000, "month": "Feb"},
    {"salesperson": "Neha", "amount": 80000, "month": "Feb"},
    {"salesperson": "Amit", "amount": 45000, "month": "Feb"},
    {"salesperson": "Neha", "amount": 90000, "month": "Mar"},
    {"salesperson": "Rohit", "amount": 55000, "month": "Mar"},
]
```

```python
def get_total_sales(data):
    total = 0
    for sale in data:
        total = total + sale["amount"]
    return total
```

```python
print(get_total_sales(sales_data))
```

```python
def get_sales_by_person(data):
    totals = {}
    for sale in data:
        name = sale["salesperson"]
        amount = sale["amount"]
        if name in totals:
            totals[name] = totals[name] + amount
        else:
            totals[name] = amount
    return totals
```

```python
print(get_sales_by_person(sales_data))
```

```python
def get_sales_by_month(data):
    totals = {}
    for sale in data:
        month = sale["month"]
        amount = sale["amount"]
        if month in totals:
            totals[month] = totals[month] + amount
        else:
            totals[month] = amount
    return totals
```

```python
print(get_sales_by_month(sales_data))
```
