```python
sales_data = [
    {"salesperson": "Rohit", "amount": 50000, "month": "Jan"},
    {"salesperson": "Amit", "amount": 75000, "month": "Jan"},
    {"salesperson": "Rohit", "amount": 60000, "month": "Feb"},
    {"salesperson": "Neha", "amount": 80000, "month": "Feb"},
    {"salesperson": "Amit", "amount": 45000, "month": "Feb"},
    {"salesperson": "Neha", "amount": 90000, "month": "Mar"},
    {"salesperson": "Rohit", "amount": 55000, "month": "Mar"},
]
```

```python
unique_months = set()
for sale in sales_data:
    unique_months.add(sale["month"])
print(unique_months)
```

```python
uniquemonths = set(sale["month"] for sale in sales_data)
print(uniquemonths)
```

```python
months_sorted = sorted(unique_months)
print(months_sorted)
```

```python
big_sellers = set()
for sale in sales_data:
    if sale["amount"] >= 60000:
        big_sellers.add(sale["salesperson"])
print(big_sellers)
```

```python
def double_strict(number):
    if not isinstance(number, (int, float)):
        raise TypeError(f"Expected a number, got {type(number).__name__}")
    return number * 2

def double_safe(number):
    try:
        return float(number) * 2
    except (ValueError, TypeError):
        return None
```

```python
print(double_strict(10))
```

```python
print(double_safe(10))
print(double_safe("10"))
print(double_safe("hello"))
print(double_safe(None))
```

```python
def get_total_sales(data):
    total = 0
    for amt in data:
        total = total + amt["amount"]
    return total
    
```

```python
results = get_total_sales(sales_data)
print(results)
```

```python
def get_sales_by_person(data):
    totals = {}
    for sale in data:
        name = sale["salesperson"]
        amount = sale["amount"]
        if name in totals:
            totals[name] = totals[name] + amount
        else:
         totals[name] = amount
    return totals
```

```python
result = get_sales_by_person(sales_data)
print(result)
```
