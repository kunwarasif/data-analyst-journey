```python
import pandas as pd
data = {
    "name": ["Rohit", "Amit", "Neha", "Vikram", "Priya"],
    "amount": [50000, 75000, 120000, 45000, 95000],
    "city": ["Delhi", "Mumbai", "Bangalore", "Chennai", "Delhi"],
    "experience_years": [3, 7, 12, 2, 9]
}
df= pd.DataFrame(data)
print(df)
```

```python
city_codes = {"Delhi": "DEL", "Mumbai": "MUM", "Bangalore": "BLR", "Chennai": "CHN"}
df["city_codes"] = df["city"].map(city_codes)
print(df)
```

```python
def categorize_amount(value):
    if value < 50000:
        return "Low"
    elif value < 100000:
        return "Medium"
    else:
        return "High"

df["amount_category"] = df["amount"].map(categorize_amount)
print(df)
```

```python
df["amount_in_lakhs"] = df["amount"].map(lambda x: x / 100000)
print(df)
```

```python
def divide_by_lakh(x):
    return x / 100000

df["amount_in_lakhs"] = df["amount"].map(divide_by_lakh)
```

```python
# Q1: Add a column "name_length" with the length of each name
df["name_length"] = df["name"].map(len)
```

```python
# Q2: Create "experience_level" based on years
#     0-3 = "Junior", 4-8 = "Mid", 9+ = "Senior"
def classify_experience(years):
    if years <= 3:
        return "Junior"
    elif years <= 8:
        return "Mid"
    else:
        return "Senior"

df["experience_level"] = df["experience_years"].map(classify_experience)
```

```python

# Q3: Use map with a dictionary to convert city to region
region_map = {"Delhi": "North", "Mumbai": "West", "Bangalore": "South", "Chennai": "South"}
df["region"] = df["city"].map(region_map)
```

```python
# Q4: Apply 18% tax to each amount using a lambda
df["amount_with_tax"] = df["amount"].map(lambda x: x * 1.18)
```

```python
# Q1: Same logic, both ways — confirm identical output
print(df["amount"].map(lambda x: round(x / 100000, 2)))
print(df["amount"].apply(lambda x: round(x / 100000, 2)))
```

```python
# Q2: apply with an extra argument — commission at a configurable rate
def commission(amount, rate):
    return amount * rate

df["commission_5pct"] = df["amount"].apply(commission, rate=0.05)
df["commission_8pct"] = df["amount"].apply(commission, rate=0.08)

```

```python
def bonus_eligibility(row):
    if row["amount"] > 70000 and row["experience_years"] >= 5:
        return "Yes"
    return "No"

df["bonus"] = df.apply(bonus_eligibility, axis=1)
print(df[["name", "amount", "experience_years", "bonus"]])
```

```python
def tiered_commission(row):
    if row["experience_years"] >= 9:
        return row["amount"] * 0.08
    elif row["experience_years"] >= 4:
        rate = 0.06 if row["amount"] > 60000 else 0.05
        return row["amount"] * rate
    else:
        return row["amount"] * 0.04

df["tiered_commission"] = df.apply(tiered_commission, axis=1)
print(df[["name", "amount", "experience_years", "tiered_commission"]])
```

```python
df["performance_flag"] = df.apply(
    lambda row: "Star" if row["amount"] / row["experience_years"] > 15000 else "Normal",
    axis=1
)
print(df[["name", "amount", "experience_years", "performance_flag"]])
```
