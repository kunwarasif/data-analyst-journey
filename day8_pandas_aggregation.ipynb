```python
def quick_report(df):
    print(f"Total Sales: ₹{df['amount'].sum():,}")
    print(f"Number of Transactions: {df['amount'].count()}")
    print(f"Average Sale: ₹{df['amount'].mean():,.0f}")
    print(f"\n--- Sales by Person ---")
    print(df.groupby("salesperson")["amount"].sum().sort_values(ascending=False))
    print(f"\n--- Sales by Month ---")
    print(df.groupby("month")["amount"].sum().sort_values(ascending=False))
    print(f"\n--- Top Performer ---")
    top = df.groupby("salesperson")["amount"].sum().idxmax()
    print(f"{top} with ₹{df[df['salesperson']==top]['amount'].sum():,}")

quick_report(df)
```
